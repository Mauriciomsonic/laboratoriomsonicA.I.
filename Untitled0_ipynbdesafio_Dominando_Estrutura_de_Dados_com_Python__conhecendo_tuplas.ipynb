```python
#!/usr/bin/env python3
import sys

def main():
    # Lê a linha de entrada e remove espaços em branco nas extremidades
    entrada = sys.stdin.readline().strip()

    # Se houver espaços, parseia como números separados por espaço;
    # caso contrário, trata cada caractere como dígito individual.
    if ' ' in entrada:
        numeros = tuple(map(int, entrada.split()))
    else:
        numeros = tuple(map(int, entrada))

    # Calcula a soma
    soma = sum(numeros)

    # Imprime no formato exato da imagem
    print(f"A soma dos elementos da tupla é: {soma}")

if __name__ == "__main__":
    main()
```
