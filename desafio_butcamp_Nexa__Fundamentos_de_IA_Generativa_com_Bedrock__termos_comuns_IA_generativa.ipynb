```python
def descrever_termo(termo):
    """
    Função que associa cada termo de IA à sua descrição correspondente
    """
    termos = {
        "Token": "Unidade de texto que modelos de linguagem processam.",
        "Fine-tuning": "Adapta um modelo treinado com dados novos e específicos.",
        "Inference": "Uso do modelo treinado para gerar saída da entrada.",
        "Dataset": "Conjunto de dados para treinar ou ajustar um modelo de IA."
    }

    return termos.get(termo, "Termo não encontrado.")

# Demonstração com exemplos
print("=== EXEMPLOS DO DESAFIO ===")

exemplos = [
    "Token",
    "Fine-tuning",
    "Inference",
    "Dataset"
]

for exemplo in exemplos:
    print(f"\nEntrada: {exemplo}")
    print(f"Saída: {descrever_termo(exemplo)}")

# Teste interativo
print("\n" + "="*50)
print("TESTE INTERATIVO")
entrada_usuario = input("Digite um termo (Token, Fine-tuning, Inference, Dataset): ").strip()
print(f"Resultado: {descrever_termo(entrada_usuario)}")
```
