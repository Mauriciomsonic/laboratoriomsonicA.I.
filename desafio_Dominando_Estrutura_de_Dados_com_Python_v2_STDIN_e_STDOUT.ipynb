```python
import sys

def elementos_comuns(lista1_str, lista2_str):
    try:
        # Converte as strings para listas de inteiros
        lista1 = list(map(int, lista1_str.split()))
        lista2 = list(map(int, lista2_str.split()))

        # Converte para conjuntos para remover duplicatas e encontrar interseção
        set1 = set(lista1)
        set2 = set(lista2)

        # Encontra os elementos comuns e converte para lista ordenada
        comuns = sorted(list(set1 & set2))

        return comuns

    except ValueError:
        return None

def main():
    # Lê as duas linhas de entrada
    linha1 = sys.stdin.readline().strip()
    linha2 = sys.stdin.readline().strip()

    # Verifica se ambas as linhas têm conteúdo
    if not linha1 or not linha2:
        print("Entrada inválida.")
        return

    # Chama a função para encontrar elementos comuns
    resultado = elementos_comuns(linha1, linha2)

    # Exibe o resultado no formato especificado
    if resultado is not None:
        print(f"Elementos comuns às duas listas: {resultado}")
    else:
        print("Entrada inválida.")

if __name__ == "__main__":
    main()

```
