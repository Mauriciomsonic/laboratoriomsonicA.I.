```python
#!/usr/bin/env python3
import sys

def soma_tupla(numeros: tuple[int, ...]) -> int:
    """
    Recebe uma tupla de inteiros e retorna a soma de seus elementos.
    """
    return sum(numeros)

def main():
    # Lê uma linha da entrada padrão e remove espaços extras
    linha = sys.stdin.readline().strip()
    # Se não houver nada na linha, consideramos soma zero
    if not linha:
        print("A soma dos elementos da tupla é: 0")
        return

    # Converte cada pedaço em int e forma uma tupla
    tupla_numeros = tuple(map(int, linha.split()))
    # Calcula a soma
    total = soma_tupla(tupla_numeros)
    # Imprime no formato esperado
    print(f"A soma dos elementos da tupla é: {total}")

if __name__ == "__main__":
    main()
```
