```python
def descrever_ferramenta(ferramenta):
    """
    Função que associa cada ferramenta AWS à sua descrição correspondente.
    """
    # Dicionário com as ferramentas e suas descrições
    ferramentas_dict = {
        "Amazon Bedrock": "Serviço de acesso via API a múltiplos modelos fundacionais.",
        "Amazon Titan": "Modelos base da AWS para gerar e resumir texto.",
        "Amazon Q": "Assistente de IA para devs e TI integrado aos serviços AWS.",
        "AMS Trainium": "Chip da AWS para treinar modelos de ML em larga escala."
    }

    # Retorna a descrição se a ferramenta existir, caso contrário retorna mensagem de erro
    return ferramentas_dict.get(ferramenta, "Ferramenta não encontrada.")

# Exemplos de uso no Colab
print("=== EXEMPLOS DE USO ===")

print("\n--- Exemplo 1 ---")
entrada1 = "Amazon Bedrock"
print(f"Entrada: {entrada1}")
print(f"Saída: {descrever_ferramenta(entrada1)}")

print("\n--- Exemplo 2 ---")
entrada2 = "Amazon Titan"
print(f"Entrada: {entrada2}")
print(f"Saída: {descrever_ferramenta(entrada2)}")

print("\n--- Exemplo 3 ---")
entrada3 = "Amazon Q"
print(f"Entrada: {entrada3}")
print(f"Saída: {descrever_ferramenta(entrada3)}")

print("\n--- Exemplo 4 ---")
entrada4 = "AMS Trainium"
print(f"Entrada: {entrada4}")
print(f"Saída: {descrever_ferramenta(entrada4)}")

# Versão interativa para teste no Colab
print("\n" + "="*60)
print("TESTE INTERATIVO")
print("="*60)
print("Ferramentas disponíveis:")
print("- Amazon Bedrock")
print("- Amazon Titan")
print("- Amazon Q")
print("- AMS Trainium")
print()

# Solicita entrada do usuário
entrada_usuario = input("Digite o nome da ferramenta AWS: ").strip()
resultado = descrever_ferramenta(entrada_usuario)
print(f"\nResultado: {resultado}")
```
