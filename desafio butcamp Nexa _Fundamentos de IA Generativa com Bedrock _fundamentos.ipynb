```python
def descrever_conceito(conceito):
    """
    Função que associa cada conceito de IA à sua descrição correspondente
    """
    # Dicionário com os conceitos e suas descrições
    conceitos_dict = {
        "Modelos de Linguagem": "IA treinada em textos para prever e gerar linguagem natural.",
        "Treinamento Supervisionado": "Aprendizado com dados rotulados para ensinar a IA a prever.",
        "Prompt Engineering": "Técnica para guiar modelos via comandos (prompts) escritos.",
        "Modelos Diffusion": "Modelo que gera imagens por refinamentos iterativos do ruído."
    }

    # Retorna a descrição se o conceito existir, caso contrário retorna mensagem de erro
    return conceitos_dict.get(conceito, "Conceito não encontrado.")

# Exemplos de uso no Colab
print("=== EXEMPLO 1 ===")
entrada1 = "Modelos de Linguagem"
print(f"Entrada: {entrada1}")
print(f"Saída: {descrever_conceito(entrada1)}")

print("\n=== EXEMPLO 2 ===")
entrada2 = "Prompt Engineering"
print(f"Entrada: {entrada2}")
print(f"Saída: {descrever_conceito(entrada2)}")

print("\n=== EXEMPLO 3 ===")
entrada3 = "Modelos Diffusion"
print(f"Entrada: {entrada3}")
print(f"Saída: {descrever_conceito(entrada3)}")

# Versão interativa para teste no Colab
print("\n" + "="*50)
print("TESTE INTERATIVO - Digite um conceito:")
print("Conceitos disponíveis: Modelos de Linguagem, Treinamento Supervisionado, Prompt Engineering, Modelos Diffusion")

# Solicita entrada do usuário
entrada_usuario = input("Digite o conceito: ").strip()
resultado = descrever_conceito(entrada_usuario)
print(f"\nResultado: {resultado}")
```
