```python
import sys

def contar_caracteres(texto: str) -> dict:
    """
    Conta a frequência de cada caractere em 'texto'
    e retorna um dicionário {caractere: quantidade}.
    """
    contador = {}
    for c in texto:
        contador[c] = contador.get(c, 0) + 1
    return contador

def formatar_saida(contador: dict) -> str:
    """
    Retorna a representação em string do dicionário de contagem.
    """
    return str(contador)

def main():
    entrada = sys.stdin.readline().strip()
    resultado = contar_caracteres(entrada)
    saida = formatar_saida(resultado)
    print(saida)

if __name__ == "__main__":
    main()
```
