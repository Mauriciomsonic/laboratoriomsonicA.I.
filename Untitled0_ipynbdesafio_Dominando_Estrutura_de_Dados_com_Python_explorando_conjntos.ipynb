```python
import sys

def elementos_comuns():
    # Lê as duas linhas de entrada
    linha1 = sys.stdin.readline().strip()
    linha2 = sys.stdin.readline().strip()

    try:
        # Converte as strings para listas de inteiros
        lista1 = list(map(int, linha1.split()))
        lista2 = list(map(int, linha2.split()))

        # Converte as listas para conjuntos para remover duplicatas
        # e encontrar a interseção entre eles
        set1 = set(lista1)
        set2 = set(lista2)
        comuns = set1 & set2

        # Converte para lista ordenada
        resultado = sorted(list(comuns))

        # Formata a saída
        print(f"Elementos comuns às duas listas: {resultado}")

    except ValueError:
        # Se ocorrer erro na conversão para inteiro
        print("Entrada inválida.")

if __name__ == "__main__":
    elementos_comuns()
```
