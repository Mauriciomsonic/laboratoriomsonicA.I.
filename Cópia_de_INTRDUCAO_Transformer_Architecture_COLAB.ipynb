```python
import requests
from bs4 import BeautifulSoup

def extraindo_texto_da_web(url):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }

    response = requests.get(url, headers=headers)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, 'html.parser')

    paragraphs = soup.find_all('p')
    text = " ".join([para.get_text() for para in paragraphs])
    return text

# URL corrigida (com maiúsculas corretas)
url = "https://pt.wikipedia.org/wiki/Microsoft_Office"
documento = extraindo_texto_da_web(url)

print(documento[:500])  # Mostra os primeiros 500 caracteres
```

```python
# Usando o Transformers

# Q & A

# Caso de Uso:
# 1. Precisamos de um modelo que entenda regras e história da Microsoft Office 365
#    responda nossas dúvidas sobre o assunto.

qa = pipeline("question-answering")

pergunta = "Qual a última versão do Office?"
result = qa(question=pergunta,
    context=documento,
    max_length=100)
print(f"Resposta: {result['answer']}\n")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline

# Carregar tokenizer e modelo explicitamente
model_name = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Criar pipeline com o modelo carregado
resumo = pipeline("summarization", model=model, tokenizer=tokenizer)

result = resumo(documento,
    max_length=100,
    min_length=50,
    do_sample=False)

print(f"Resumo: {result[0]['summary_text']}\n")
```

```python

# Configuração
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Carregar modelo e tokenizer
tokenizer = AutoTokenizer.from_pretrained('mrm8488/bert-mini2bert-mini-finetuned-cnn_daily_mail-summarization')
model = EncoderDecoderModel.from_pretrained('mrm8488/bert-mini2bert-mini-finetuned-cnn_daily_mail-summarization').to(device)

def resumo(documento):
    inputs = tokenizer([documento], padding="max_length", truncation=True, max_length=512, return_tensors="pt")
    input_ids = inputs.input_ids.to(device)
    attention_mask = inputs.attention_mask.to(device)

    output = model.generate(input_ids, attention_mask=attention_mask, max_length=150, min_length=30)

    return f"Resumo: {tokenizer.decode(output[0], skip_special_tokens=True)}"

# Usar com o documento existente
if 'documento' in dir() or 'documento' in locals():
    print(resumo(documento))
else:
    print("Variável 'documento' não encontrada. Forneça um texto para resumir.")
```
