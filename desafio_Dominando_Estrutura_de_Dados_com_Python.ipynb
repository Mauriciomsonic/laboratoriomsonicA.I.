```python
import sys

def main():
    # Simulando os dados de entrada dos testes
    testes = [
        "2 5 6 7 9",    # Teste #1 - soma deve ser 29
        "9 8 7 6 5",    # Teste #2 - soma deve ser 35
        "50 50 50 50"   # Teste #3 - soma deve ser 200
    ]

    for i, entrada in enumerate(testes, 1):
        print(f"Teste #{i}")
        print(f"Entrada: {entrada}")

        # Converte os números para inteiros e cria uma tupla
        numeros = tuple(map(int, entrada.split()))

        # Calcula a soma dos elementos
        soma = sum(numeros)

        # Exibe o resultado no formato especificado
        print(f"Saída: A soma dos elementos da tupia é: {soma}")
        print("-" * 40)

if __name__ == "__main__":
    main()

```
